```python
import pymatgen as mg
import os
from natsort import natsorted
from pymatgen.io.vasp.inputs import Poscar
from pymatgen.symmetry.analyzer import SpacegroupAnalyzer
import re
```

```python
path = 'C:/Users/Adam/Desktop/voro-ml-si/datasets/icsd-all/'
fileList = natsorted(os.listdir(path))
```

```python
i=0
j=0
total = fileList.__len__()
for file in fileList:
    j=j+1
    if os.path.splitext(file)[1]=='':
        s = mg.Structure.from_file(path+file)
        if s.ntypesp==2:
            if mg.Composition(s.formula).anonymized_formula=='AB2':
                formula = str(i)+'-'+str.replace(s.formula,' ','')
                s = Poscar(s)
                s.write_file('icsd-AB2/'+formula+'.POSCAR')
                i=i+1
                if i%1000==0:
                    print(str(i)+' files identified and converted.')
                    print(str(round(j/total*100,1))+'% database scan completed.')
```

```python
for file in ab2List:
    s = mg.Structure.from_file('oqmd-AB2/'+file)
    fractions = re.sub(r'[A-z]+', '', s.formula).split()
    elements = re.sub(r'[0-9]+', '', s.formula).split()
    if int(fractions[0])/int(fractions[1])==2:
        s.replace_species({mg.Element(elements[0]): mg.Element('Bi')})
        s.replace_species({mg.Element(elements[1]): mg.Element('Nd')})
    elif int(fractions[0])/int(fractions[1])==0.5:
        s.replace_species({mg.Element(elements[1]): mg.Element('Bi')})
        s.replace_species({mg.Element(elements[0]): mg.Element('Nd')})
    else:
        print('Error. Not AB2.')
    # Analyze space group
    spGroupA = SpacegroupAnalyzer(s)
    fileSplit = file.split('.')
    newName = fileSplit[0]+'-'+spGroupA.get_crystal_system()+'-'+str(spGroupA.get_space_group_number())+'-'+str.replace(s.formula,' ','')+'.'+fileSplit[1]
    sP = Poscar(s)
    sP.write_file('oqmd-AB2-NdBi2/'+newName)
    sIdeal = spGroupA.get_refined_structure()
    newName2 = fileSplit[0]+'-ideal_'+spGroupA.get_crystal_system()+'-'+str(spGroupA.get_space_group_number())+'-'+str.replace(s.formula,' ','')+'.'+fileSplit[1]
    sP = Poscar(sIdeal)
    sP.write_file('oqmd-AB2-NdBi2-ideal/'+newName2)
```
